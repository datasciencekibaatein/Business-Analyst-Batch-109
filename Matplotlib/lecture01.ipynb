```python
import numpy as np
import pandas as pd
```

```python


np.random.seed(42)

df = pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","oct","Nov","Dec"],
    "Sales":np.random.randint(100,300,12),
    "Profit":np.random.randint(15,30,12),
    "Customers":np.random.randint(5,30,12),
    "Marketing":np.random.randint(10,25,12),
})

df['Profit_Margin'] = df['Profit']/df['Sales'] * 100
```

```python
df
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Month'],df['Sales'])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹ Thousands)")

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(11,5))

ax.plot(
    df['Month'],
    df['Sales'],
    color="#2563EB",
    linewidth=3,
    linestyle="-",
    marker="o",
    markersize=12,
    markerfacecolor="black",
    markeredgecolor="red",
    markeredgewidth=1,
    label="Sales"
    )

ax.set_title("Monthly Sales",fontsize=22,color="#2563EB",fontweight="bold")
ax.set_xlabel("Month",fontsize=12)
ax.set_ylabel("Sales (₹ Thousands)",fontsize=12)

ax.grid(axis="x",alpha=0.3,linestyle="--")
ax.legend()
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,5))

ax.plot(df['Month'],df['Sales'],marker='o',label='Sales')
ax.plot(df['Month'],df['Profit'],marker='s',label='Profit')
ax.plot(df['Month'],df['Marketing'],marker='^',label='Marketing')

ax.set_title("Sales, Profit and Marketing")
ax.set_xlabel("Month")
ax.set_ylabel("Amount (₹ thousand)")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,5))

ax.bar(df['Month'],df['Sales'])

ax.set_title("Monthly sales Comparision")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,5))

bars = ax.bar(
    df['Month'],
    df['Sales'],
    color='teal',
    width=0.7,
    edgecolor='black',
    linewidth=0.3
    )

ax.bar_label(bars,fmt='%.1f%%',padding=3,fontsize=9)

ax.set_title("Monthly sales Comparision",fontsize=16,fontweight='bold')
ax.grid(axis='y',alpha=0.25)
ax.set_axisbelow(True)
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,5))

bars = ax.barh(df['Month'],df['Sales'],color='magenta')
ax.bar_label(bars,padding = 4)
ax.set_title("Sales comparision")
ax.set_ylabel("Month")
ax.set_xlabel("sales")

plt.show()
```

```python
x = np.arange(len(df['Month']))
width = 0.38

fig,ax = plt.subplots(figsize=(12,5))
ax.bar(x - width/2,df['Sales'],width,label='Sales')
ax.bar(x + width/2, df['Profit'],width,label='Profit')

ax.set_xticks(x)
ax.set_xticklabels(df['Month'])

ax.set_title("Sales vs Profit")
ax.set_ylabel("Amount (₹ thousand)")
ax.legend()

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,5))

ax.bar(df['Month'],df['Profit'],label="Profit")
ax.bar(
    df['Month'],
    df['Sales']-df['Profit'],
    bottom=df['Profit'],
    label="Remaining Sales"
)

ax.set_title("Profit and Remaining Sales")
ax.legend()

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8,5))

ax.scatter(
    df['Marketing'],
    df['Sales'],
    color='#8B5CF6',
    s=80,
    alpha=0.8,
    linewidths=0.5
)

ax.set_title("Marketing Spend vs Sales")
ax.set_xlabel("Marketing (₹ Thousands)")
ax.set_ylabel("Sales (₹ Thousands)")

ax.grid(alpha=0.3)

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(8,5))

sc = ax.scatter(
    df['Marketing'],
    df['Sales'],
    c=df['Profit'],
    s=df['Customers']*2,
    cmap='viridis',
    alpha=0.8,
    linewidths=0.5
)
fig.colorbar(sc, ax=ax, label="Profit (₹ Thousands)")
ax.set_title("Marketing Spend vs Sales")
ax.set_xlabel("Marketing (₹ Thousands)")
ax.set_ylabel("Sales (₹ Thousands)")

ax.grid(alpha=0.3)

plt.show()
```

```python
scores = np.random.normal(loc=70,scale=10,size=1000)
scores = np.clip(scores,0,100)

fig,ax = plt.subplots(figsize=(9,5))

ax.hist(scores,bins=10,color='#F59e0b',edgecolor='white')

ax.set_title("Distribution of Exam Scores")
ax.set_xlabel("Score")
ax.set_ylabel("Frequency")

plt.show()
```

```python
group_a = np.random.normal(65,10,500)
group_b = np.random.normal(75,8,500)


fig,ax = plt.subplots(figsize=(9,5))

ax.hist(group_a,bins=20,alpha=0.55,label='Group A')
ax.hist(group_b,bins=20,alpha=0.55,label='Group B')

ax.set_title("Scores distribution by group")
ax.set_xlabel("score")
ax.set_ylabel("Frequency")

ax.legend()

plt.show()
```

```python
scores = np.random.normal(loc=70,scale=10,size=1000)
scores = np.clip(scores,0,100)

fig,ax = plt.subplots(figsize=(9,5))

ax.hist(scores,bins=10,color='#F59e0b',edgecolor='white',density=True)

ax.set_title("Probability density of Exam Scores")
ax.set_xlabel("Score")
ax.set_ylabel("Frequency")

plt.show()
```

```python
salary = np.random.normal(50000,12000,500)
salary = np.append(salary,100000,120000)

fig,ax = plt.subplots(figsize=(8,5))
ax.boxplot(
    salary,
    showmeans=True,
    showfliers=True,
    patch_artist=True,
)
ax.set_title("Salary Distribution")
ax.set_ylabel("Salary")
plt.show()
```

```python
department_salaries = [
    np.random.normal(50000,8000,500),
    np.random.normal(60000,10000,500),
    np.random.normal(45000,7000,500)
]


fig,ax = plt.subplots(figsize=(8,5))
ax.boxplot(
    department_salaries,
    tick_labels=['Sales','Finance','Tech'],
    showmeans=True,
    showfliers=True,
    patch_artist=True,
)
ax.set_title("Salary Comparision by departments")
ax.set_ylabel("Salary")
plt.show()
```

```python
channels = ['Website','Retail','Referral','Social Media']
revenue = [45,25,20,10]

fig,ax = plt.subplots(figsize=(7,7))

ax.pie(
    revenue,
    autopct='%1.1f%%',
    labels=channels,
    startangle=180,
    shadow=True,
    explode=[0.1,0,0,0],
    labeldistance=1.1,
    colors=['teal','magenta','cyan','blue']

)

ax.set_title("Revenue by channels")
plt.show()
```
