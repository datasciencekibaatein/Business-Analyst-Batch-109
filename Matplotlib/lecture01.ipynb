# Matplotlib



```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3],[10,25,36])
plt.show()
```

```python
# object oriented approach

fig,ax = plt.subplots()
ax.plot([10,20,30],[34,20,45])
```

Figure - The entire canvas containing the visulization.

Axes - The individual plotting area, including labels, title and plotted data.

Axis - The x-axis or y-axis, responsible for ticks, tick and scale

```python
fig,ax = plt.subplots(figsize = (8,5))

ax.plot([1,2,3,4,5],[200,300,340,380,500])
ax.set_title("Monthly sales")
ax.set_xlabel("Months")
ax.set_ylabel("Sales (₹ Thousand)")
plt.show()
```

```python
import numpy as np
import pandas as pd
```

```python


np.random.seed(42)

df = pd.DataFrame({
    "Month":["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","oct","Nov","Dec"],
    "Sales":np.random.randint(100,300,12),
    "Profit":np.random.randint(15,30,12),
    "Customers":np.random.randint(20,30,12),
    "Marketing":np.random.randint(10,25,12),
})

df['Profit_Margin'] = df['Profit']/df['Sales'] * 100
```

```python
df
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Month'],df['Sales'])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹ Thousands)")

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(11,5))

ax.plot(
    df['Month'],
    df['Sales'],
    color="#2563EB",
    linewidth=3,
    linestyle="-",
    marker="o",
    markersize=12,
    markerfacecolor="black",
    markeredgecolor="red",
    markeredgewidth=1,
    label="Sales"
    )

ax.set_title("Monthly Sales",fontsize=22,color="#2563EB",fontweight="bold")
ax.set_xlabel("Month",fontsize=12)
ax.set_ylabel("Sales (₹ Thousands)",fontsize=12)

ax.grid(axis="x",alpha=0.3,linestyle="--")
ax.legend()
plt.show()
```
