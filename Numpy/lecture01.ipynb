```python
import numpy as np
```

```python
arr = np.array([34,23,45,56,34])
print(arr)
```

```python
lst = [34,89,78,49]
arr1 = np.array(lst)


print("Numpy array: ",arr1)
print("Standard python list: ",lst)
print("Numpy array type: ",type(arr1))
print("Type of list: ",type(lst))
```

```python
arr1 * 2
```

```python
import time


num = 10_00_000
# [0 - 999999] x 2

lst = list(range(num))

start = time.time()
print(start," start the list operation")
result = [x * 2 for x in lst]
end = time.time()
print(end," end the list operation")
print("List operation time: ",(end - start)*100)


arr = np.arange(num)
start = time.time()
print(start," start the np operation")
result = arr * 2
end = time.time()
print(end," end the np operation")
print("Numpy array time: ",(end - start)*100)



```
