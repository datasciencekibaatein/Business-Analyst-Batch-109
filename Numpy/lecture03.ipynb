# Operations on np arrays

```python
import numpy as np
```

```python
arr = np.array([67,23,83,20])

```

```python
num = 4

res = arr + num

print(res)
```

```python
res = num * arr
print(res)
```

```python
res = arr // 4
print(res)
```

```python
matrix = np.array(
    [
        [3,2,5],
        [7,8,9],
        [9,3,2]
    ]
)

print(matrix * num)
```

If we have any scalar value and if you perform operation using . so each element of the matrix or array going to be the part of that operation(broadcast).
