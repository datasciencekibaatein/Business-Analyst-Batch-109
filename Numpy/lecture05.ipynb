## Dataset

- 100 employees
- 6 months
- 4 region
- 5 products

# Numerical Data

```python
n = 100

employee_id = np.arange(1001,1001 + n)
age = np.random.randint(22,55,n)
experience = np.random.randint(0,16,n)
sales = np.random.randint(20000,150000,n)
units_sold = np.random.randint(20,100,n)
discount = np.round(np.random.uniform(0,30,n),2)
customer_rating = np.round(np.random.uniform(2.5,5,n),2)
```

## Categorical Data

```python
regions = np.random.choice(
    ["North","South","East","West"],
    n
)

products = np.random.choice(
    ["Laptop","Mobile","Tablet","Monitor","Printer"],
    n
)
```

```python
regions
```

```python
products
```

```python
data = np.zeros(n,
dtype=[
    ("Employee_id",'i4'),
    ("Age",'i4'),
    ("Experience",'i4'),
    ("Region","U10"),
    ("Product","U15"),
    ("Sales","f8"),
    ("Unit_sold","i4"),
    ("Discount","f8"),
    ("Customer_Rating","f8")
])
```

Now populate data:

```python
data['Employee_id'] = employee_id
data['Age'] = age
data['Experience'] = experience
data['Region'] = regions
data['Product'] = products
data['Sales'] = sales
data['Unit_sold'] = units_sold
data['Discount'] = discount
data['Customer_Rating'] = customer_rating
```

```python
regions,counts = np.unique(data['Region'],return_counts=True)
```

```python
for region,count in zip(regions,counts):
    print(f"{region} : {count}")
```

```python
data['Sales'][data['Region']=="West"]
```

```python
for region in regions:
    region_sales = data['Sales'][data['Region']==region]
    print("-"*5,f"{region}' stat","-"*5)
    print(f"Total sale: {np.sum(region_sales)}")
    print(f"Average sale: {np.round(np.mean(region_sales),3)}")
    print(f"Minimum sale: {np.min(region_sales)}")
    print(f"Maximum sale: {np.max(region_sales)}")
    print()
    print()


```

```python
prdcts,prdcnt = np.unique(data['Product'],return_counts=True)
print(prdcts,prdcnt)

for prd in prdcts:
    product_sales = data['Sales'][data['Product']==prd]
    print("-"*5,f"{prd}' stat","-"*5)
    print(f"Total sale: {np.sum(product_sales)}")
    print(f"Average sale: {np.round(np.mean(product_sales),3)}")
    print(f"Minimum sale: {np.min(product_sales)}")
    print(f"Maximum sale: {np.max(product_sales)}")
    print()
    print()
```
