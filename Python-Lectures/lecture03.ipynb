Input function

```python
age = int(input("Enter the age: "))

if age >= 18 and age <=100:
    print("You can vote.")

else:
    print("You can't vote.")
```

### Nested statement

```python
salary = int(input("Enter the salary: "))
if salary >=30000:#outer if statement
    age = int(input("Enter the age: "))

    if age >= 18 and age <=100:#inner if statment
        print("You can vote.")

    else:
        print("You can't vote.")

else:
    print("Sorry, Your salary is not greater than 30000")
```

### if-elif-else

Marks evaluation

```python
marks = int(input("Enter the marks: "))

if marks >=0 and marks <=100:
    if marks>=90:
        print("A Grade")
    elif marks >=80:
        print("B Grade")
    elif marks >=70:
        print("C Grade")
    elif marks >=60:
        print("D Grade")
    elif marks >=33:
        print("E Grade")
    else:
        print("Fail")
else:
    print("Enter valid number.")
```

01. Check if number is positive or negative.

```python
num = int(input("Please enter the number: "))
if num != 0:
    if num>0:
        print (f"Number {num} is positive")
    else:
        print (f"Number {num} is negative")
else:
    print(f"Number {num} is zero")
```

2. Check if the number is even or odd.

```python
num = int(input("enter a number:"))
if num%2==0:
    print("the number is even")
else:
    print("the number is odd")
```

```python
num1 = int (input("Please enter first  number: "))
num2 = int (input("Please enter second  number: "))
if num1 == num2:
    print ("Both numbers are equal")
elif num1 > num2:
    print (f"first {num1} number is greater")
else: 
    print (f"second {num2} number is greater")
```
