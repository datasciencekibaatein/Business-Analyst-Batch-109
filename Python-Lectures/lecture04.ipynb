# Set

A set stores unique value and does not maintain the orderd(unordered datastructure)

to create set , use the curly braces

```python
set1 = {5,6,2,2,3,5,8}
set2 = {5,7,8,3,0,1}
print(set1)
print(set2)
```

##### Functions in set

```python
#add function will add the element in set , if element already exist, it will not be added,(no change is set, it wil remain same)
set1.add(9)
set1.add(9)
print(set1)
```

```python
# it will remove the element from the set if exists else it will throw error
set1.remove(9)
print(set1)
```

```python
# it is similar to remove but it does not throw error , if element doesn't exists otherwise it removes the element.
set1.discard(8)
print(set1)
```

```python
# it will remove all the elements from the set
set1.clear()
print(set1)
```

###### Set operations

```python
print(set1)
print(set2)
```

```python
#union

print(set1.union(set2))
```

```python
# intersection
# common elements from given sets

print(set1.intersection(set2))
```

```python
# difference 

print(set1.difference(set2))
```

```python
print(set2.difference(set1))
```
