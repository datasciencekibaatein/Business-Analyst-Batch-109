```python
num  = 6
res = 1

for i in range(1,num+1):
    print(f"res = res : {res} * i: {i}")
    res = res * i
    print("res: ",res)


print(res)
```

```python
num  = 5
res = 1

for i in range(num,1,-1):
    print(f"res = res : {res} * i: {i}")
    res = res * i
    print("res: ",res)


print(res)
```

```python
def factorial(num):
    res = 1
    for i in range(num,1,-1):
        res = res * i
    
    return res

  
```

```python
factorial(5)
```

```python
factorial(6)
```

## WAP to sum of digits of a given number e.g. {34523}

```python
num = int(input("Enter the number: "))

res = 0

while num > 0:
    re = num % 10
    res = res + re
    num//=10

print(res)
```

```python
456//10
```

```python
def digits_sum(num):
    res = 0

    while num > 0:
        re = num % 10
        
        res = res + re
        num//=10

    print(res)

    
```

```python
digits_sum(782324)
```

### Armstrong number (67345)

- length = 5(power)

Armstrong number where the sum of the power(length of number) of each number is equal the same number.

67345 = 6^5 + 7^5 + 3^5 + 4^5 + 5^5


examples of armstrong numbers:

1634, 8208, and 9474
