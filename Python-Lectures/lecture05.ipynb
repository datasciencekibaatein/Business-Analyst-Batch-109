# Dictionary

A dictionary (dict) in Python is a collection that stores data as key-value pairs.

```python
employee = {
    "name":"Rahul",
    "age":30,
    "salary":56000,
    "department":"IT",
    "name":"Rahul"
}
```

```python
print(employee)
print(type(employee))
```

```python
employee["name"]
```

```python
employee["age"]
```

```python
print(employee['id'])
```

### Add pair in dictionary

```python
print(employee)
employee["email"] = "rahul@example.com"
print(employee)
```

### Updating data in dictionary

```python
print(employee["email"])
employee["email"] = "rahul@gmail.com"
print(employee)
```

```python
print(employee['id'])
```
