### WAP using only loop

1. Reverse a number using loop.
2. Count the number of digits in a number.
3. Find the sum of digits of a number.
4. Find the factorial on a number.

```python
num = int(input("Enter the number: "))

rev = 0

while num > 0:
    re = num % 10
    rev = rev * 10 + re
    num//=10


print("Reverse: ",rev)
```
