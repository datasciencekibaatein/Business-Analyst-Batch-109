# Encapsulation

Encapsulating (wrapping) the code and object in the class.

- public
- default
- protected
- private

```python
class BankAccount:
    def __init__(self,name,amount,city):
        self.name = name
        self.__amount = amount #private
        self.city = city

    def detail(self):
        return f"Name: {self.name} \nBalance: {self.__amount}\nCity: {self.city}"
```

```python
rohan_acc = BankAccount("Rohan",34000,"Delhi")
```

```python
print(rohan_acc.detail())
```

```python
rohan_acc.amount-= 30000
```

```python
rohan_acc.name
```

# Inhertance

```python
class Vehicle:
    def __init__(self,name):
        self.name = name

    def v_func(self):
        print(f"I am vehicle class. {self.name}")

class Car(Vehicle):
    def c_func(self):
        print(f"I am car class. {self.name}")
    
    def start(self):
        print(f"I am starting the {self.name}")

class Trucks(Vehicle):
    def t_func(self):
        print(f"I am truck class. {self.name}")

class Sedan(Car):
    def s_func(self):
        print(f"I am sedan class {self.name}")

    def start(self):
            print(f"Starting {self.name}")

class Convertible(Car):
    def cv_func(self):
        print(f"I am Convertible class {self.name}")

    def start(self):
            print(f"I am starting the {self.name}")
```

```python
v = Vehicle("Demo")
v.v_func()
```

```python
c = Car("Skoda")
c.c_func()
```

```python
sdn = Sedan("Verna")

sdn.start()
```

```python
cvt = Convertible("Cyberster")
cvt.start()
```

# Abstraction

```python
from abc import ABC,abstractmethod

class Employee(ABC):

    @abstractmethod
    def work(self):
        pass

class Manager(Employee):
    def work(self):
        print("Manager assigns tasks to employees")
    
class Developer(Employee):
    def work(self):
        print("Developer writes and tests code")


manager = Manager()
developer = Developer()

manager.work()
developer.work()
```

```python
class Student:
    def __init__(self, name, roll):
        self.name = name
        self.roll = roll
        self.marks = []

    def add_mark(self, mark):
        self.marks.append(mark)

    def get_average(self):
        if self.marks:
            return sum(self.marks) / len(self.marks)
        return 0

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll}")
        print(f"Marks: {self.marks}")
        print(f"Average: {self.get_average():.2f}")


# Driver code
students = []

while True:
    print("\n1. Add Student\n2. Show Students\n3. Exit")
    choice = input("Enter choice: ")

    if choice == "1":
        name = input("Enter name: ")
        roll = input("Enter roll no: ")
        s = Student(name, roll)
        for i in range(3):
            mark = int(input(f"Enter mark {i+1}: "))
            s.add_mark(mark)
        students.append(s)

    elif choice == "2":
        for s in students:
            s.display_info()

    elif choice == "3":
        break

    else:
        print("Invalid choice")
```
