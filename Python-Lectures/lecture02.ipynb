### Arithmetic Operators

Used to perform mathematical Operations

|Operator | Meaning | Example| Output|
|-----|-----|-----|---|
|+ | Addition | 10 + 5 |15|
| "-" | Subtraction | 10 - 5 | 5 |
| * | Multiplication | 10 * 5 | 50 |
| / | Division | 10/2 | 2.0 |
| // | floor division | 10//3 | 3|
| % | Modulus(Remainder) | 10 % 3 | 1|
| ** | Exponent | 2**3 | 8 |

```python
a = 19
b = 3

print(f"Addition: {a + b}")
print(f"Subtraction: {a - b}")
print(f"Multiplication: {a * b}")
print(f"Divsion: {a / b}")
print(f"Floor Division: {a // b}")
print(f"Modulus(Remainder): {a % b}")
print(f"Exponent: {a ** b}")


```

### Comparision Operators

Return either True or False(Boolean values).


|Operator  | Meaning |
|------|------|
|== | Equal to|
|!= | Not equal to|
|> | Greater Than |
|< | Less Than |
|>= | Greater Than or equal to |
|<= | Less Than or equal to |

```python
a = 45
b = 45

print(f"Equal to : {a == b}")
print(f"Not equal to : {a != b}")
print(f"Greater Than : {a > b}")
print(f"Less than : {a < b}")
print(f"Greater than or equal to : {a >= b}")
print(f"Less than or equal to : {a <= b}")

```

### Logical Operators

|Operator | Meaning |

```python
age = 20
salary = 56000
department ="IT"
```

### Membership Operators

Check a value exists in a sequence.

|Operator | Meaning|
|-----|-----|
| in | Present |
| not in | Not Present |

```python
lst = [23,34,12,45,65]
print(45 in lst)
print(13 not in lst)
```

## Identity Operators

### Bitwise Operators

|Operator  | Meaning |
|------|------|
| & | AND|
| ` | OR|
| ^ | XOR |
| ~ | NOT |
| << | Left Shift |
| >>| Right Shift|

```python
num1 = 12
num2 = 10
print(num1 & num2)
print(num1 | num2)
print(num1 ^ num2)
print(~num1) #-(n+1)
print(num1<<2)
print(num1>>3)
```
