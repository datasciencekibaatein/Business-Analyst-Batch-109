# Iterative Statments

We can use iterative statments.. if we want to execute the same statements again and again .

In range() function stop point always will be excluded

```python
for i in range(12):#range(stop) : when you will pass the stop point iteration will start from 0
    print(i)
```

in range(start,stop): start point will be included and stop will be excluded

```python
for i in range(3,11):#range(start,stop) : it will start iteration from {start} point and go till {stop} point
    print(i)
```

```python
for i in range(2,21,3):#range(start,stop,step)
    print(i)
```

```python
for i in range(30,1,-2):
    print(i)
```

```python
fruits = ["Guava","apple","Papaya"]

for fruite in fruits:
    print(fruite)
```

```python
lst = [34,56,78,23,98,67]

for num in lst:
    print(num)
```

```python
lst = [34,56,78,23,98,67]

for num in lst:
    if num % 2==0:
        print("Even: ", num)
    else:
        print("Odd: ",num)
```

# Iteration in Dictionary

```python
st = {
    "name":"Neha",
    "Age":22,
    "course":"BSc",
    "city":"Delhi"
}
```

```python
st.keys()
```

```python
for key in st.keys():
    print(key)
```

```python
st.values()
```

```python
for value in st.values():
    print(value)
```

```python
st.items()
```

```python
for tp in st.items():
    print(tp)
```

```python
st.items()
```

```python
for key,value in st.items():
    print(f"{key} : {value}")
```

In loop 

```python
# Wap to generate the list of even number in range of 1 to 100

even_lst = []

for num in range(1,101):
    if num % 2 == 0:
        even_lst.append(num)

print(even_lst)
```

```python
even_lst = []

for num in range(2,101,2):
    even_lst.append(num)

print(even_lst)
```

```python
evn_lst = [num for num in range(2,101,2)]
print(evn_lst)
```

```python
evn_lst = [num for num in range(1,101) if num % 7==0]
print(evn_lst)
```
