```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

```

```python
df = pd.read_csv("merged_table.csv")
```

```python
df.head()
```

```python
df.columns
```

## Scatter Plot

```python
plt.figure(figsize=(10,5))
sns.scatterplot(
    data=df,
    x="Profit",
    y="Sales",
    hue="Region_x",
    palette="rocket",
    alpha=0.5
)
plt.title("Profit VS Sale",fontweight="bold",fontsize=22)
plt.xlabel("Profit",fontsize=12)
plt.ylabel("Sale",fontsize=12)
plt.legend(loc="lower right")
plt.show()
```

## Line Plot

```python
flights = sns.load_dataset("flights")

plt.figure(figsize=(10,5))

sns.lineplot(
    data=flights,
    x='year',
    y='passengers',
    marker="o",
    errorbar=None
)
plt.title("Passenger trend Over time")
plt.show()
```

```python
df = sns.load_dataset("tips")
```

```python
df.head()
```

## Relational Plot

```python
plt.figure(figsize=(10,5))
sns.relplot(
    data=df,
    x="total_bill",
    y="tip",
    kind='scatter',
    col='sex',
    row='smoker'
)

plt.show()
```

## histogram

```python
plt.figure(figsize=(10,5))
sns.histplot(
    data=df,
    x= "total_bill",
    bins=20,
    color='steelblue',
    kde=True
)
plt.axvline(df['total_bill'].mean(),color='red',label=f"Mean: {df['total_bill'].mean():.2f}")
plt.axvline(df['total_bill'].median(),color='green',label=f"Median: {df['total_bill'].median():.2f}")

plt.title("bill amount distribution")
plt.xlabel("Bill amount")
plt.ylabel("frequency")
plt.legend(loc='upper right')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.kdeplot(
    data=df,
    x= "total_bill",
    color='teal',
    fill=True,
    alpha=0.3
)
plt.axvline(df['total_bill'].mean(),color='red',label=f"Mean: {df['total_bill'].mean():.2f}")
plt.axvline(df['total_bill'].median(),color='green',label=f"Median: {df['total_bill'].median():.2f}")

plt.title("bill amount Density")
plt.xlabel("Bill amount")
plt.ylabel("Probability")
plt.legend(loc='upper right')
plt.show()
```

## ECDF (Empirical Cumulative Distribution Function) Plot

```python
plt.figure(figsize=(10,5))

sns.ecdfplot(
    data=df,
    x='total_bill',
    hue='smoker'
)

plt.show()
```

## Count plot

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x='day',
    hue='sex',
    palette="Set2"
)
plt.title("customer count by day & Gender")

plt.show()
```
