## Join

Join combines the dataframe based on their index.

```python
df1 = pd.DataFrame(
    {
        "Name":["Amit","Riya","Rahul"],
        "Department":["IT","HR","Finance"],
    },index=[101,102,103]
)
```

```python
df2 = pd.DataFrame(
    {
        "Salary":[50000,60000,45000]
    },index = [101,102,104]
)
```

```python
df1.join(df2)
```

```python
df1.join(df2,how='left')
```

```python
df1.join(df2,how='right')
```

```python
df1.join(df2,how='inner')
```

```python
df1.join(df2,how='outer')
```

```python
df3 = pd.DataFrame(
    {   "EmpID":[101,102,103],
        "Name":["Amit","Riya","Rahul"],
        "Department":["IT","HR","Finance"],
    }
)
```

```python
df2
```

```python
df3
```

```python
df3.join(df2,on='EmpID')
```

### Difference between join() and merge()

|Feature | join() | merge() |
|--------|-------|--------|
|Default matching key | index | common columns, when available |
|Column based joining | supported with on parameter | directly supported with on |
|Best Suited for | Joining  data frames based on index | Joining tables using key columns|
