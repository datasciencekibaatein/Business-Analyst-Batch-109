```python
import pandas as pd
```

```python
df = pd.read_csv("products.csv",sep="$")
```

```python
df.head()
```

## Value counts

```python
df['Category'].value_counts()
```

## Group by

```python
df.groupby("Category")
```

```python
df['Category'].value_counts()
```

```python
df.groupby("Category")['Unit_Cost'].mean()
```

```python
df.groupby("Category")['Unit_Cost'].min()
```

```python
df.groupby("Category")['Unit_Cost'].sum()
```

```python
df.groupby("Category")['Unit_Cost'].max()
```

```python
df.groupby("Category")['Unit_Cost'].count()
```

```python
df.groupby("Category")['Unit_Cost'].median()
```

```python
df.groupby("Category")['Unit_Cost'].std()
```

```python
df.groupby("Category")['Unit_Cost'].var()
```

#### Multiple aggregations

```python
df.groupby('Category')['Unit_Cost'].agg(['mean','sum','count'])
```

#### Group multiple columns

```python
df.groupby(['Category','Subcategory'])['Unit_Cost'].mean().reset_index()
```

```python
df.groupby('Category')[['Unit_Cost','List_Price']].mean()
```

```python
df.groupby('Category').agg(
   { "Unit_Cost":'sum',
    "List_Price":"mean",
    "List_Price":"sum"}
)
```

```python
df.groupby('Category').agg(
   { "Unit_Cost":'sum',
    "List_Price":["mean","sum"],
    }
)
```

#### Named aggregation

```python
df.groupby('Category').agg(
   Avg_Cost = ("Unit_Cost","mean"),
    Avg_Price = ("List_Price","mean")
)
```

## Crosstab

crosstab tells frequency/count

```python
df.head()
```

```python
pd.crosstab(df['Category'],df['Subcategory'])
```

```python
pd.crosstab(df['Category'],df['Subcategory'],normalize='index')
# each row sum to 100%
```

```python
pd.crosstab(df['Category'],df['Subcategory'],normalize='columns')

# each column sum to 100%
```

```python
pd.crosstab(df['Category'],df['Subcategory'],margins=True)
```

```python
pd.crosstab(df['Category'],df['Subcategory'],values=df['Unit_Cost'],aggfunc='mean')
```

## pivot table

it is like excel pivot table

```python
pd.pivot_table(
    df,
    index = "Subcategory",
    columns = "Category",
    values=['Unit_Cost','List_Price'],
    fill_value=0,
    aggfunc=["sum",'mean']
)
```
