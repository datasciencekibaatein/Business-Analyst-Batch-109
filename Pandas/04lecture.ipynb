```python
import pandas as pd
```

```python
orders = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Orders")
order_items = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Order_Items")
customers = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Customers")
products = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Products")
categories = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Categories")
stores = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Stores")
salepersons = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Salespersons")
returns = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Returns")
targets = pd.read_excel("Indian_Sales_Dataset.xlsx",sheet_name="Targets")
```

## merge() function

```python
pd.merge(orders,customers,how='inner',on="Customer_ID").head()
```

```python
pd.merge(orders,customers,how='left',on="Customer_ID").head()
```

```python
pd.merge(orders,customers,how='right',on="Customer_ID").head()
```

```python
pd.merge(orders,customers,how='outer',on="Customer_ID").head()
```

```python
sales = pd.DataFrame(
    {
        "Store":['Delhi','Delhi','Mumbai'],
        "Products":['Laptop','Mouse','Laptop'],
        "Sales":[50000,1000,60000]
    }
)


trgt = pd.DataFrame(
    {
        "Store":['Delhi','Delhi','Mumbai'],
        "Products":['Laptop','Mouse','Laptop'],
        'target':[55000,1200,65000]
    }
)


pd.merge(sales,trgt,on=["Store","Products"])
```

```python
sales = pd.DataFrame(
    {
        "Store":['Delhi','Delhi','Mumbai'],
        "products":['Laptop','Mouse','Laptop'],
        "Sales":[50000,1000,60000]
    }
)


trgt = pd.DataFrame(
    {
        "store":['Delhi','Delhi','Mumbai'],
        "Products":['Laptop','Mouse','Laptop'],
        'target':[55000,1200,65000]
    }
)

pd.merge(sales,trgt,left_on=["Store","products"],right_on=['store','Products'])
```

```python
df =pd.merge(
    pd.merge(
        pd.merge(
            pd.merge(
            pd.merge(orders,customers,on='Customer_ID',how='inner'),
            salepersons,on='Salesperson_ID',how='inner'),
                pd.merge(targets,stores,how='inner',on='Store_ID'),
                how='inner',on='Store_ID'),
                returns,on='Order_ID',how='inner'),
                pd.merge(pd.merge(order_items,products,on='Product_ID',how="inner")
                ,categories,on='Category_ID',how='inner'),how='inner',on='Order_ID')
```
